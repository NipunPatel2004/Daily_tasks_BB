```python
import pandas as pd

from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import RobustScaler
```

```python
df = pd.read_csv("Admission_Prediction.csv")

df.head()
```

```python
# select num data to scale 
num = df.select_dtypes(include="number")
print(num.head())
```

```python

#select the input feature and remove the target col
x = num.drop("Chance of Admit", axis=1)
# remove unwanted columns 
x = num.drop("Serial No.", axis=1)
```

```python
# min max scaler 
minmax = MinMaxScaler()
m_result = minmax.fit_transform(x)
df_minmax = pd.DataFrame(m_result,columns=x.columns)

df_minmax.head()
```

```python
# robust scaler 
robust = RobustScaler()

r_result = robust.fit_transform(x)
robust_df = pd.DataFrame(r_result,columns=x.columns)
print(robust_df.head())
```

```python
standard = StandardScaler()

result_s = standard.fit_transform(x)
standard_df = pd.DataFrame(result_s,columns=x.columns)
print(standard_df.head())
```
